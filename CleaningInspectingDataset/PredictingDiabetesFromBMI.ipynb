```python
import pandas as pd
import numpy as np

#In this project, I inspect and clean a dataset for diabetes. This allows me to conduct a logistic regression, whereby we can predict the probability
#of diabetes diagonsis from BMI! 


diabetes_data = pd.read_csv("diabetes.csv")
```

```python
print(diabetes_data.head(10))
```

```python
#to find out if any columns is missing data
print(diabetes_data.isnull().sum())
```

```python
#another method to see if data is missing any values, which finds minimums of 0 (missing data) for variables, and questionable maximums e.g 17 for pregnancy 
print(diabetes_data.describe())
```

```python
#replaces instances of 0 with NaN, to exclude these values from our summary statistics 
diabetes_data[['Glucose','BloodPressure','SkinThickness','Insulin','BMI']] = diabetes_data[['Glucose','BloodPressure','SkinThickness','Insulin','BMI']].replace(0,np.nan)
```

```python
#prints rows which have missing (null) values. Every row with at least one missing value, has a missing value in the insulin column

print(diabetes_data[diabetes_data.isnull().any(axis=1)])
```

```python
#Making sure columns are expected data types 
print(diabetes_data.dtypes)
#Determining why Outcome column is an object type instead of int64, which appears to be caused by inputs of "O" instead of "0"
print(diabetes_data.Outcome.unique())
```

```python
# Replace 'O' with 0 in the Outcome column
diabetes_data['Outcome'] = diabetes_data['Outcome'].replace('O', 0)

# Convert the Outcome column to int64
diabetes_data['Outcome'] = diabetes_data['Outcome'].astype('int64')

```

```python
#now we have cleaned the data, we can predict the probability of one being diabetic, given their BMI 

import statsmodels.api as sm
import matplotlib.pyplot as plt 

#lets remove all NAN values before we conduct our logistic regression, otherwise we will get an error 
diabetes_data1 = diabetes_data.dropna() 

#we can use logistic regression to do this, since we have a binary variable (outcome for diabetes) and a continious variable (BMI). 
#this model shows that for each unit increase in BMI, the probability that one is diabetic increases by 0.017! (Z = 5.06, very sig!)

X = diabetes_data1[["BMI"]]
X = sm.add_constant(X)
Y = diabetes_data1["Outcome"]
model = sm.Logit(Y, X)
result = model.fit()
print(result.summary())


#lets plot our model on a graph, so we can clearly see the relationship 

x_vals = np.linspace(diabetes_data1["BMI"].min(), diabetes_data1["BMI"].max(), 100)
X_plot = pd.DataFrame({"intercept": 1, "BMI": x_vals})
y_vals = result.predict(X_plot)


plt.scatter(diabetes_data1["BMI"], diabetes_data1["Outcome"], label = "Actual Data")
plt.plot(x_vals, y_vals, color = "red", label = "Logistic fit")
plt.xlabel("BMI")
plt.ylabel("Probability of Diabetes")
plt.legend()
plt.title("Logistic Regression Curve")
plt.show()






```
