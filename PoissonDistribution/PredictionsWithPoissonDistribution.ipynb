```python
import scipy.stats as stats
import numpy as np

#Imagine the rate of defects a day at a factory, is 7, which follows a Poisson Distribution 
lam = 7 

# What is the probability that we will get exactly 7 defects? 
print(stats.poisson.pmf(lam, lam))

# What is the probability, that we will get 4 or fewer defects? ,We use a cumulative density function to do this! 
print(stats.poisson.cdf(4, lam))

# What is the probability that we have more than 9 defects on a given day?
print(1 - stats.poisson.cdf(9, lam))
```

```python
# What is the probability that we have more than 9 defects on a given day?
print(1 - stats.poisson.cdf(9, lam))

# Lets use this Poission distribution, to generate random values for each day of the year 
year_defects = stats.poisson.rvs(lam, size = 365)
print(year_defects)

# If we had an average rate of 6 errors per day, we would expect 2555 errors over the year 
print(7*365)

# Adding our random data set, which could be more representive of the real world than the previous multiplication, is 2659/  
print(year_defects.sum())

#  Moreover, the average errors per day from our random dataset, is 7.28!  
print(year_defects.mean())

# This randomly generated dataset following the Poission distribution, will allow us to prepare for a plausible maximum amount of errors in a single day! (16)
print(year_defects.max())


# We can then calculate the probability of this day happening using the probability mass function! 
print(stats.poisson.pmf(16, lam))

# On 90% of days, our amount of errors will be this amount or lower!
percentile = 0.9
print(stats.poisson.ppf(percentile, lam))

# We can see that 19 percent of days, are greater or equal to this number! 
print(sum(year_defects >= 10)/len(year_defects))


```
